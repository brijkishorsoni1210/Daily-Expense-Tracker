**Project title: Daily Expense Tracker**


**Problem statement :** We need to design a python console application that use for calculate the expenses which a user had made on that day. The application able to do following task:

*   Total expense
*   Average expense
*   Highest expense







**Logic:**


1.   Design the function to take user input for daily expenses
2.   Store them in a list
3.   Convert to a Numpy array
4.   Perform the Analysis to calcualte Total expense, Average expense and Highest expense
5.   Provide the list of expenses in summarize view
6.   Program complete







# Program

```python
import numpy as np
def analyze_expenses():
  dates=[]
  expenses=[]
  print("-----------Daily expenses--------------")
  while True:
    day_input=input("Enter the Date and Day (e.g.'25 Oct, Monday') or type 'done' to finish:")
    if day_input.lower()=='done':
      break
    try:
      amount=float(input(f"Enter expense for {day_input}:$"))
      dates.append(day_input)
      expenses.append(amount)
    except ValueError:
      print("Invalid input. Please enter a numeric value for the expense")
  if not expenses:
    print("No expenses entered. Exiting program...")
    return
  expenses_array=np.array(expenses)

  #Total expense
  total_expenses=np.sum(expenses_array)

  #Average expense
  Average_expense=np.mean(expenses_array)

  #Highest expense
  Highest_expense=np.max(expenses_array)
  highest_date=dates[np.argmax(expenses_array)]

  print("\n=====Spending Summary=====\n")
  print(f"\n{'Date':<25} | {'Expense($)':<10}")
  print("-"*39)
  for i in range(len(dates)):
    print(f"{dates[i]:<25} | {expenses[i]:<10.2f}")

  print("-"*39)
  print(f"Total Expense:   ${total_expenses:.2f}")
  print(f"Average Expense: ${Average_expense:.2f}")
  print(f"Highest Expense: ${Highest_expense:.2f} on {highest_date}")

  budget_limit=300.00
  if Average_expense >budget_limit:
    print(f"\n[WARNING] Your average daily expense (${Average_expense:2f}) exceeds your limit")

if __name__ =="__main__":
  analyze_expenses()
```
